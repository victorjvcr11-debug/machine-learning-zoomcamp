```python
#Q4. Missing values
incomplete_columns = df.columns[df.isnull().any()].tolist()

print(incomplete_columns)
```

```python
#Q5. Max fuel efficiency
asian_cars = df[df['origin'] == 'Asia']
max_val = asian_cars['fuel_efficiency_mpg'].max()
print('The maximun value of fuel efficiency is', max_val)
asian_cars
```

```python
#Q6. Median value of horsepower
mean_before = df['horsepower'].mean()
print(f'The mean before fill missing values is {mean_before:.2f}')
mode = df['horsepower'].mode()[0]
print(f'The most frequent value is', mode)

df['horsepower'] = df['horsepower'].fillna(mode)
mean_after = df['horsepower'].mean()
print(f'The mean after fill missing values is {mean_after:.2f}')
df['horsepower']
if mean_before == mean_after:
    print('The mean doesnt change.')
elif mean_before > mean_after:
    print('The mean decrease.')
else:
    print('The mean increase.')
```

```python
#Q7. Sum of weights
asian_cars = df[df['origin'] == 'Asia']
d_filter = asian_cars[['vehicle_weight', 'model_year']].head(7)

x = d_filter.to_numpy()

XTX = np.dot(x.T, x)

XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = np.dot(np.dot(XTX_inv, x.T), y)

sum_w = np.sum(w)

print("Weights vector is:", w)
print("The sum of all the elements is:", sum_w)
```
